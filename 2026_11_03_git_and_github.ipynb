## What are Git and GitHub?

**Git** is a **version control system** designed to let you keep track of possibly multiple versions of your code, with no chance of accidentally deleting or losing anything.
It also helps a **team of coders** keep track of different features being tried out for a code base, helping to merge them together after they have been debugged.
Git keeps a **repository of your files, including all saved versions, on your local computer.**

**GitHub** is a commercial, online **Git Server**, owned by Microsoft since 2018.  It provides a cloud-based place to save Git repositories, and a "social network" for teams working on them. Many well-known projects are developed and maintained on GitHub, including:
- VS Code
- Ubuntu
- Pytorch
- Most open-source software projects

Knowing how to use Git and GitHub are important skills that many research groups and potential employers will be happy to see you have.
A couple of more detailed references:
- As I was first learning how to use Git and GitHub, I wrote a very detailed "cheat sheet" that you can find [here](https://github.com/doncandela/gs-git) (on GitHub).
- A standard reference book on Git is [Pro Git by Chacon and Straub](https://git-scm.com/book/en/v2).

The `git commit` command makes an **un-alterable snapshot** of your staged files, which can always be restored in the future.
This is the basic function of Git!

## A few handy Git terminal commands

See your local branches (you will have only `main` for now):
```
$ git branch -vv
```
See your commit history (there are many options to `git log` to show things the way you want):
```
$ git log --oneline    # gives on commit per line
```
Stage all files for commit (used above), or stage a specific file for commit:
```
$ git add .
$ git add <filename>
```
Commit all staged files, or stage and commit all tracked files (`git add` must be done at least once start tracking a file):
```
$ git commit -m 'my commit message'
$ git commit -am 'my commit message'
```
I find this last command to be the handiest way to "snapshot" my repo after making edits or other changes.
Note all commits must have a message -- if you don't supply one an editor will open for you to write it.


## Reverting (undoing) changes to your files

Use the **nano** editor to make a file we can play around with:
```
$ nano foofile.txt    # then enter some contents and save the file.
```
Set this new file to be tracked, then make a new commit:
```
$ git add foofile.txt      # only needed once for each new file
$ git commit -am 'added foofile.txt'  # stages and commits
```
Make some changes to the file and commit again:
```
$ nano foofile.txt   # then make some changes and save
$ git commit -am 'changed foofile.txt'
$ git log --oneline  # see the commit history
```
Now we decide that the changes were a mistake, and we want to go back to the previous version of `foofile.txt`.
Rather than "erasing" the most recent commit (which can be done but is discouraged and made difficult on purpose) it is better to **revert** it, which means doing a new commit that reverses the changes (leaving the previous, changed version available):
```
$ git revert HEAD     # will open editor for message, can accept proposed
$ git log --oneline
```
The version with the changes is still available for us to **check out.** (Don't worry about the "Detached HEAD" message, it's not as bad as it sounds.)
```
$ git checkout <hash of version with changes>
$ cat foofile.txt        # will see version with changes
$ git checkout main      # back to the latest (reverted) version
$ cat foofile.txt        # will see latest version
```
A Git checkout actually **changes the files on our laptop** to match the checked-out commit.
For this reason you should **never use git along with some other type of file syncing** (like OneDrive, Google Drive, iCloud Drive) on the same directory.

## **Carrying on**

**Using Git + GitHub to back up and (as needed) revert your files.**  The simplest, yet quite handy use of Git + GitHub is to keep well-organized backup copies as you write your code, with the ability to go back to earlier versions if the need arises.
- If you create **new files** in your repo, you must add them to Git tracking (only once, as Git remembers which files are tracked):
    ```
    $ git add <filename>   # add a specific new file
    $ git add .            # all all new files in or under current directory
    ```
- These commands stage all tracked files for commit, commit them to the local Git repo, then push the changes to GitHub. After doing these commands you should see your latest files if you login to your GitHub account:
    ```
    $ git status       # not necessary but always good to see what is going on
    $ git commit -am 'my commit message'
    $ git push         # push to GitHub
    ```
- As already shown above your can do a new commit that **undoes (in a reversible way) the most recent commit**:
    ```
    $ git revert HEAD -m 'my message for this reversion'
    $ git revert HEAD --no-edit      # alternative uses default message
    $ git push            # push to GitHub
    ```
- Here's how to do a commit that **puts your files back to an earlier (than the most recent) commit**, again in a reversible way.  First you must find the **hash code** for the commit you want to go back to. Do this by looking at the commit history, either locally by doing
    ```
    $ git log --oneline
    ```
  or on GitHub, where you can click on the **history** icon (looks like a clock), click on any commit, then inspect the files as they were in that commit.  Next do
    ```
    $ git checkout <hash code> -- .
    $ git commit -m 'my message eg rolling back to ...'
    $ git push            # push to GitHub
    ```
  Note the added parts `-- .` on the `git checkout` command above - these tell Git to simpy change your files to match the earlier commit without creating a "Detached HEAD", so these changes are ready to commit.  Here the '.' means all files in the current directory and subdirectories, so do this command while at the top of your repo if you want to roll back the entire repo.

- You can use all of the steps purely locally in Git, without using GitHub at all -- in this case you would omit the `git push` commands.  The disadvantages are (a) if your laptop is lost or damaged you would lose your files, (b) you cannot work on your files from another laptop or PC as described below and (c) you cannot so easily share your files with someone else by pointing them to your GitHub repo. Finally, as a physics student you should know how to use GitHub!

**Working from more than one laptop or PC.**  Assuming you have pushed your files to GitHub, they can now be accessed and worked on from any PC or laptop.
- Any other PC or laptop you want to use must be configured with `git config` commands and set to authenticate to GitHub as described above.
- To grab a repo from GitHub (your own repo, or someone else's repo if it is public), go to the repo on github.com and click the green **<> Code** button.  Use the dropdown to copy the HTTPS address of the repo (or the SSH address, if you are using SSH keys to authenticate).  Independent of what OS you used to create the repo and push it to GitHub, you can clone it to a machine using any other OS (macOS, Windows+WSL, Windows without WSL, Linux).
- Go to the the directory on the laptop or PC *under which* you want to create a **clone** (working copy) of the repo, and in this directory do
    ```
    $ git clone <copied HTTPS or SSH address>
    ```
  Unless you are using SSH keys or you are cloning a public repo, you will be asked to authenticate to GitHub.
- Now you can work in this copy of the repo (which has the same status in all respects as your original repo -- all clones are equal).
- It will be simplest if you push all changes made to GitHub with `git push`, and then always use `git pull` on any other laptop or PC before making further changes on that other computer.
- Git and GitHub can certainly deal more complicated situations, like compatible or incompatible changes made to a repo on two different computers withouth pushing and pulling as above.  See [https://github.com/doncandela/gs-git](https://github.com/doncandela/gs-git) for into on these more interesting situations.
